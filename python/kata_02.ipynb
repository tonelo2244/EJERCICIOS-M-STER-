Dada una lista de números, obtén una nueva lista con el doble de cada valor. Usa la función map().

```python
# Creamos una lista de números
numeros = [5, 10, 20]

# Creamos una función que devuelve el doble de un número
def duplicar(numero):
    return numero * 2

# Aplicamos la función duplicar a cada número de la lista
resultado = list(map(duplicar, numeros))

# Mostramos el resultado
print(resultado)
```
