```python
#Crea una función que convierta una variable en una cadena de texto y enmascare todos los caracteres con el carácter '#' excepto los últimos cuatro.
def enmascarar_excepto_ultimos_cuatro(valor):
    texto = str(valor)
    return "#" * (len(texto) - 4) + texto[-4:]


print(enmascarar_excepto_ultimos_cuatro(1234567890123456))
print(enmascarar_excepto_ultimos_cuatro("ABC"))
```
