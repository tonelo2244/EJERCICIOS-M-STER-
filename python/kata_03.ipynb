 Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros. La función debe devolver una lista con todas las palabras de la lista original que contengan la palabra objetivo.

```python
#Creamos una funcion que recibe una lista de palabras 
# y una palabra objetivo qu queremos buscar 
def buscar_palabras(lista_palabras, objetivo):

    # Creamos una lista vacía donde guardaremos
    # las palabras que coincidan
    resultado = []

# Recorremos una por una todas las palabras
    for palabra in lista_palabras:

# Si la palabra objetivo está dentro de la palabra actual
        if objetivo in palabra:

    
# Añadimos esa palabra a resultado
            resultado.append(palabra)


    # Devolvemos todas las palabras encontradas
    return resultado


# Lista de ejemplo
palabras = ["freno", "frenada", "motor", "frenillo", "rueda"]


# Buscamos las palabras que contienen "fren"
print(buscar_palabras(palabras, "fren"))
```
