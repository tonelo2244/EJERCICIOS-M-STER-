```python
#Escribe un programa que pida al usuario que introduzca su edad. 
# Si el usuario ingresa un valor no numérico o un valor fuera del rango esperado (por ejemplo, menor que 0 o mayor que 120), 
# maneja las excepciones adecuadamente.

EDAD_MINIMA = 0
EDAD_MAXIMA = 120


class EdadFueraDeRangoError(Exception):
    pass


def pedir_edad():
    entrada = input("Introduce tu edad: ")
    edad = int(entrada)
    if edad < EDAD_MINIMA or edad > EDAD_MAXIMA:
        raise EdadFueraDeRangoError(f"La edad debe estar entre {EDAD_MINIMA} y {EDAD_MAXIMA}")
    return edad


try:
    edad = pedir_edad()
    print(f"Tu edad es: {edad}")
except ValueError:
    print("Error: debes introducir un número entero")
except EdadFueraDeRangoError as error:
    print(f"Error: {error}")
```
