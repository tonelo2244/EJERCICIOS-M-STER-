```python
#Escribe un programa que pida al usuario dos números e intente dividirlos. Si el usuario ingresa un valor no numérico o intenta dividir por cero, 
#maneja esas excepciones de manera adecuada y muestra un mensaje indicando si la división fue exitosa o no.
```

```python
try:
    numero1 = float(input("Introduce el primer número: "))
    numero2 = float(input("Introduce el segundo número: "))

    resultado = numero1 / numero2

    print("División realizada correctamente.")
    print(resultado)

except ValueError:
    print("Debes introducir números.")

except ZeroDivisionError:
    print("No se puede dividir entre cero.")
```
