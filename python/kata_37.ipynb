```python
#Genera un programa que nos indique si es de noche, de día o de tarde según la hora proporcionada por el usuario.
# Antonio lo mismo que el ejercicio anterior me ha costado 
hora = int(input("Introduce una hora entre 0 y 23: "))

if hora >= 6 and hora < 12:
    print("Es de día")

elif hora >= 12 and hora < 20:
    print("Es de tarde")

elif hora >= 0 and hora < 24:
    print("Es de noche")

else:
    print("Hora no válida")
```
