```python
#Genera una función que calcule la diferencia entre los valores de dos listas. 
# Usa la función map().

# Función que calcula la diferencia entre dos valores
def restar(a, b):
    return a - b


# Dos listas de ejemplo
lista1 = [10, 20, 30]
lista2 = [3, 5, 10]


# Aplicamos la función restar a las dos listas
resultado = list(map(restar, lista1, lista2))


# Mostramos el resultado
print(resultado)
```
