```python
#Genera una función que, para un conjunto de caracteres, 
# devuelva una lista de tuplas con cada letra en mayúsculas y minúsculas. Las letras no pueden estar repetidas. Usa la función map().

def crear_tupla_mayuscula_minuscula(letra):
    return (letra.upper(), letra.lower())


def obtener_letras_en_ambos_casos(caracteres):
    letras_unicas = sorted(set(caracteres.lower()))
    return list(map(crear_tupla_mayuscula_minuscula, letras_unicas))


caracteres = "PythonPYTHON"
resultado = obtener_letras_en_ambos_casos(caracteres)
print(resultado)
```
