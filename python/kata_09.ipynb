```python
#Escribe una función que tome una lista de nombres de mascotas como parámetro y devuelva una nueva lista excluyendo ciertas mascotas prohibidas en España. 
# La lista de mascotas a excluir es ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]. Usa la función filter().
# Antonio esto no sale en temario, está corregido con ia 

MASCOTAS_PROHIBIDAS = ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]


def es_mascota_permitida(mascota):
    return mascota not in MASCOTAS_PROHIBIDAS


def filtrar_mascotas_permitidas(mascotas):
    return list(filter(es_mascota_permitida, mascotas))


mis_mascotas = ["Perro", "Gato", "Tigre", "Loro", "Oso", "Hámster"]
mascotas_permitidas = filtrar_mascotas_permitidas(mis_mascotas)
print(mascotas_permitidas)
```
