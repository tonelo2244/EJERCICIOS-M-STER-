```python

# Escribe una función que tome dos parámetros: figura (una cadena que puede ser "rectangulo", "circulo" o "triangulo") 
# y datos (una tupla con los datos necesarios para calcular el área de la figura).


import math


def calcular_area(figura, datos):

    if figura == "rectangulo":
        base = datos[0]
        altura = datos[1]
        return base * altura

    if figura == "circulo":
        radio = datos[0]
        return math.pi * radio ** 2

    if figura == "triangulo":
        base = datos[0]
        altura = datos[1]
        return base * altura / 2

    return "Figura no válida"


print(calcular_area("rectangulo", (5, 3)))
print(calcular_area("circulo", (4,)))
print(calcular_area("triangulo", (6, 4)))
```
