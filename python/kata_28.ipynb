```python
#Crea una función que busque y devuelva el primer elemento duplicado en una lista dada.
def encontrar_primer_duplicado(elementos):
    vistos = set()
    for elemento in elementos:
        if elemento in vistos:
            return elemento
        vistos.add(elemento)
    return None


elementos = [3, 7, 1, 9, 7, 3, 5]
print(encontrar_primer_duplicado(elementos))
```
